
**26. Connect and create a table (5 pts)**

- (a) Import `sqlite3` and connect to a database named `'school.db'`. Save the connection in a variable named `conn` and print `"Opened database successfully"`.
- (b) Run this line first so the cell can be re-run without errors (it deletes the old table if it already exists):
  ```python
  conn.execute('DROP TABLE IF EXISTS STUDENTS')
  ```
- (c) Create a table named `STUDENTS` with these columns, then print `"Table created successfully"`:

| Column | Type | Rule |
|---|---|---|
| `ID` | `INT` | `PRIMARY KEY`, `NOT NULL` |
| `NAME` | `TEXT` | `NOT NULL` |
| `AGE` | `INT` | `NOT NULL` |
| `COURSE` | `CHAR(50)` | |
| `GWA` | `REAL` | |

````python
import sqlite3

conn = sqlite3.connect('school.db')
print("Opened database successfully")

conn.execute('DROP TABLE IF EXISTS STUDENTS')

conn.execute('''
CREATE TABLE STUDENTS (
    ID INT PRIMARY KEY NOT NULL,
    NAME TEXT NOT NULL,
    AGE INT NOT NULL,
    COURSE CHAR(50),
    GWA REAL
)
''')
print("Table created successfully")
````

| ID | NAME | AGE | COURSE | GWA |

| 2 | Ben | 20 | BSIT | 2.25 |


````python
conn.execute("INSERT INTO STUDENTS (ID, NAME, AGE, COURSE, GWA) VALUES (1, 'Ana', 19, 'BSCS', 1.50)")
conn.execute("INSERT INTO STUDENTS (ID, NAME, AGE, COURSE, GWA) VALUES (2, 'Ben', 20, 'BSIT', 2.25)")
conn.execute("INSERT INTO STUDENTS (ID, NAME, AGE, COURSE, GWA) VALUES (3, 'Carla', 19, 'BSCS', 1.75)")
conn.execute("INSERT INTO STUDENTS (ID, NAME, AGE, COURSE, GWA) VALUES (4, 'Dan', 21, 'BSIS', 2.50)")
conn.commit()
print("Records created successfully")

cursor = conn.execute("SELECT ID, NAME, COURSE, GWA FROM STUDENTS")
for row in cursor:
    print("ID = ", row[0])
    print("NAME = ", row[1])
    print("COURSE = ", row[2])
    print("GWA = ", row[3], "\n")
````
